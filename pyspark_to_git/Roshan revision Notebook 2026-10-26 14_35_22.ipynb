```python

import csv

with open("/tmp/Customer_details.csv", "r", encoding="utf-8") as file:
    reader = csv.DictReader(file)
    rows = list(reader)

df = spark.createDataFrame(rows)

display(df)
```

```python
df.printSchema()
df.show(10, truncate=False)
print("Total rows:", df.count())
```

A company wants to identify customers from Yangon whose age is greater than 25.

```python
from pyspark.sql.functions import col
df_filtered = df.filter(
    (col("city") == "Yangon" )&
    (col("age").cast("int")>25)
)
result = df_filtered.select(
    "age","City","Customer type","Gender","Payment"
)
result.show(truncate = False)
print("total matching customer",result.count())
```

The company wants to find customers who use Ewallet and give a rating of 8 or higher.

```python
from pyspark.sql.functions import col
df_filltred =df.filter(
    (col("Payment")=="Ewallet")&
    (col("Rating").cast("int")>=8)
)
result = df_filltred.select(
    "Age","City","Payment","Rating","Quantity"
) 
result.show(truncate = False)
print("total matching customer ",result.count())
```

The company wants to classify customers into age groups for reporting.

Using your existing df DataFrame:
1. Create a new column named Age_Group.
2. If age is less than 25, assign "Young".
3. If age is 25 or greater, assign "Adult".
4. Display Age, City, and Age_Group.
5. Show the first 20 rows.

```python
from pyspark.sql.functions import col,when
df_select = df.withColumn(
    "Age_group",
    when(col('Age').cast("int")<25,"Young")
    .otherwise("Adult")
)
result = df_select.select(
    "Age","City","Age_group"
)
result.show(20)
```

A company wants to classify premium customers.

Using your existing df DataFrame:
1. Create a column named Customer_Category.
2. If Salary >= 30000 and Rating >= 8, assign "Premium".
3. Otherwise, assign "Regular".
4. Display City, Salary, Rating, and Customer_Category.
5. Show the first 20 rows.

```python
from pyspark.sql.functions import col ,when
df_column = df.withColumn(
    "Customer_Category",
    when(
    (col("salary").cast("int")>=30000)&
    (col("Rating").cast("int")>= 8),
    "Premium"
    ).otherwise("Regular")   
)
result = df_column.select(
    "City","Salary","Rating","Customer_Category"
)
result.show(20,truncate = False)
```
